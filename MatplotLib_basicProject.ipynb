```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Set seed for reproducibility
np.random.seed(42)

data = {
    'Date': pd.date_range(start='2026-01-01', periods=10, freq='D'),
    'online': np.random.randint(20, 80, size=10),
    'physical': np.random.randint(50, 100, size=10),
    'expenses': np.random.randint(10, 30, size=10)
}
sample_data = pd.DataFrame(data)

sample_data.head()
```

```python
from matplotlib.lines import lineStyles
# Write your code below
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(sample_data['Date'],sample_data['online'],marker='o',linestyle='-', color='r',label='Online Store Sales')
plt.plot(sample_data['Date'],sample_data['physical'],marker='o',linestyle='-', color='b',label='Physical Store Sales')
plt.plot(sample_data['Date'],sample_data['online'].rolling(window=3).mean(),linestyle='--', color='pink',label='Online Sales - MA')
plt.plot(sample_data['Date'],sample_data['physical'].rolling(window=3).mean(),linestyle='--', color='skyblue',label='Physical Sales - MA')

plt.xlabel('Dates')
plt.ylabel('Sales for Online and Physical')
plt.legend()
plt.grid(True)
plt.title('Sales Over Time with Moving Average')

plt.show()
```

```python
plt.figure(figsize=(12,5))
width=0.35

plt.bar(sample_data['Date']-pd.Timedelta(days=width/2),sample_data['online'],width=width,label='Online Store Sales', color='r')
plt.bar(sample_data['Date']+pd.Timedelta(days=width/2),sample_data['physical'],width=width,label='Physical Store Sales', color='b')

plt.ylabel('Sales for Online and Physical')
plt.xlabel('Date')
plt.legend()
plt.title('Sales Comparison - Grouped Bar Chart')
```
